### Models Integration wuth OpenAi, Google Gemini and GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv

os.environ["OPENAI_API_KEY"]=os.getenv("open_ai_api_key")
os.environ["GOOGLE_API_KEY"]=os.getenv("google_api_key")
os.environ["GROQ_API_KEY"]=os.getenv("groq_api_key")



```

```python
from langchain.chat_models import init_chat_model
models=init_chat_model("gpt-4o")

models
```

```python
## invoke the model

response=models.invoke("Hello How are you ?")
response.content
```

### google gemini model integration

```python
import os 
from langchain.chat_models import init_chat_model
os.environ["GOOGLE_API_KEY"]=os.getenv("google_api_key")
model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response=model.invoke("why do parrot talk ?")
response.content
```

### chat open AI

```python
from langchain_openai import ChatOpenAI
model1=ChatOpenAI(model="gpt-4o")
response = model1.invoke("How are you ?")
response.content
```
